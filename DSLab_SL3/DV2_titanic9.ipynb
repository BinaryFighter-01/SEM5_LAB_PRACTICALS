1. Use the inbuilt dataset 'titanic' as used in the above problem. Plot a box plot for distribution of age with respect to each gender along with the information about whether they survived or not. (Column names : 'sex' and 'age')
2. Write observations on the inference from the above statistics.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the Titanic dataset
titanic = sns.load_dataset('titanic')
```

```python
print(titanic.head())
```

```python
# Plot a boxplot for age distribution by gender and survival status
plt.figure(figsize=(10, 6))
sns.boxplot(x='sex', y='age', hue='survived', data=titanic)
# Add title and labels
plt.title('Age Distribution by Gender and Survival Status on the Titanic')
plt.xlabel('Gender')
plt.ylabel('Age')

# Display the plot
plt.show()
```

```python
# Observations (printed for clarity)
print("""
Observations from the Box Plot:
1. **Age Distribution by Gender**:
   - For males, the median age is similar for both survivors (around 28–30) and non-survivors (around 30–32).
   - For females, the median age of survivors (around 25–28) is slightly lower than non-survivors (around 30–35).
   - The age range (interquartile range, IQR) for males is wider than for females, indicating more variability in male ages.

2. **Survival Patterns**:
   - Among males, survivors tend to be slightly younger than non-survivors, though the difference is small.
   - Among females, survivors are noticeably younger on average than non-survivors, with a tighter age distribution for survivors.
   - Outliers (extreme ages) are present in both genders, with some older passengers (age > 60) surviving, particularly among females.

3. **Gender and Survival**:
   - Females show a clearer distinction in age between survivors and non-survivors compared to males.
   - The plot suggests younger females had a higher survival rate, possibly due to the 'women and children first' evacuation policy.
   - Males have more outliers in the non-survivor group, indicating older males were less likely to survive.

4. **General Insights**:
   - The box plot highlights that age alone doesn't fully explain survival, but younger females had a survival advantage.
   - The presence of outliers suggests some elderly passengers survived, which could be due to specific circumstances (e.g., class, status, or luck).
""")
```
