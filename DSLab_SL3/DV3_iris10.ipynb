```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Step 1: Load and scan the Iris dataset
df = pd.read_csv('iris.csv')
```

```python
# Show the first few rows of the dataset
print(df.head())   
```

```python
# 2. Plot histograms for each feature (except species)
df[['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']].hist(bins=30)
plt.show()
```

```python
# 3. Plot boxplots for each feature
sns.boxplot(df[['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']])
plt.show()
```

```python
print("\nObservations:")
print("1. **Sepal Length**:")
print("   - Distribution: Roughly normal, slightly right-skewed, peaks around 5.0-6.0 cm.")
print("   - Outliers: Few, possibly values above 7.5 cm or below 4.5 cm.")
print("   - Range: ~4.3 to 7.9 cm, IQR ~5.1 to 6.4 cm.")
print("2. **Sepal Width**:")
print("   - Distribution: Normal, centered around 3.0 cm, slight spread.")
print("   - Outliers: Notable outliers below 2.5 cm and above 4.0 cm (e.g., ~4.4 cm).")
print("   - Range: ~2.0 to 4.4 cm, IQR ~2.8 to 3.3 cm.")
print("3. **Petal Length**:")
print("   - Distribution: Bimodal, with peaks around 1.5 cm (Setosa) and 4.5-5.5 cm (Versicolor/Virginica).")
print("   - Outliers: None significant, as the spread aligns with class separation.")
print("   - Range: ~1.0 to 6.9 cm, IQR ~1.6 to 5.1 cm.")
print("4. **Petal Width**:")
print("   - Distribution: Bimodal, peaks at ~0.2 cm (Setosa) and 1.5-2.0 cm (Versicolor/Virginica).")
print("   - Outliers: None significant, consistent with class differences.")
print("   - Range: ~0.1 to 2.5 cm, IQR ~0.3 to 1.8 cm.")
print("\nComparison:")
print("- Sepal length and width are more normally distributed, with sepal width showing more outliers.")
print("- Petal length and width are bimodal, reflecting the distinct Iris species (Setosa vs. others).")
print("- Outliers are most prominent in sepal width, indicating variability not tied to class.")
print("- Petal measurements have wider ranges but fewer outliers, due to clear species separation.")
```
