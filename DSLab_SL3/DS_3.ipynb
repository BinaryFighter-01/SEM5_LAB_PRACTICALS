Part 1: Summary Statistics Grouped by Categorical Variable

```python
# Import required libraries
import pandas as pd
```

```python
# Step 1: Load the dataset
df = pd.read_csv('iris.csv')
print("First 5 rows of the dataset:")
print(df.head())
```

```python
# 1. Group by 'Species' (Capital 'S')
grouped = df.groupby('Species')
```

```python
# Summary statistics
summary_stats = grouped.describe()  
print("\nSummary Statistics (grouped by Species):")
print(summary_stats)
```
